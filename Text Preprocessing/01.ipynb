### **Removng URL**

```python
def removeURL(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
import string, time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def removePunctuation(text):
    for char in exclude:
        text = text.replace(char,'')
    return text
```

```python
text = "Hello, , my name is PRAjwal? and I'am a data scientiest"
```

```python
start = time.time()
print(removePunctuation(text))
time1 = time.time() - start
print(time1)
```

```python
def removePunctuation1(text):
    return text.translate(str.maketrans('','',exclude))
```

```python
start = time.time()
print(removePunctuation1(text))
time2 = time.time() - start
print(time2)
```

```python
def chatConverion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```
